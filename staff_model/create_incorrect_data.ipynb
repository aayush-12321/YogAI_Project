# Important Functions

```python
import pandas as pd
import numpy as np
import random

IMPORTANT_LANDMARKS = [
    "nose",
    "left_shoulder", "right_shoulder",
    "left_elbow",    "right_elbow",
    "left_wrist",    "right_wrist",
    "left_hip",      "right_hip",
    "left_knee",     "right_knee",
    "left_ankle",    "right_ankle",
    "left_heel",     "right_heel",
    "left_foot_index","right_foot_index",
]

def get_landmark_columns(landmark_name):
    return [
        f'{landmark_name}_x',
        f'{landmark_name}_y',
        f'{landmark_name}_z',
        f'{landmark_name}_v'
    ]

def modify_landmark(landmarks, landmark_name, x_delta=0, y_delta=0, z_delta=0):
    modified = landmarks.copy()
    modified[f'{landmark_name}_x'] += x_delta
    modified[f'{landmark_name}_y'] += y_delta
    modified[f'{landmark_name}_z'] += z_delta
    return modified


def create_rounded_back(landmarks):
    """Spine rounds — shoulders collapse forward and down, hips tilt back."""
    modified = landmarks.copy()
    round_amount  = np.random.uniform(0.06, 0.14)
    forward_shift = np.random.uniform(0.03, 0.08)

    modified = modify_landmark(modified, 'nose',
        y_delta=round_amount * 0.8, x_delta=forward_shift)
    modified = modify_landmark(modified, 'left_shoulder',
        y_delta=round_amount, x_delta=forward_shift * 0.8)
    modified = modify_landmark(modified, 'right_shoulder',
        y_delta=round_amount, x_delta=forward_shift * 0.8)
    modified = modify_landmark(modified, 'left_elbow',
        y_delta=round_amount * 0.6, x_delta=forward_shift * 0.5)
    modified = modify_landmark(modified, 'right_elbow',
        y_delta=round_amount * 0.6, x_delta=forward_shift * 0.5)
    modified = modify_landmark(modified, 'left_wrist',
        y_delta=round_amount * 0.4)
    modified = modify_landmark(modified, 'right_wrist',
        y_delta=round_amount * 0.4)
    # Hips tilt back — posterior pelvic tilt
    modified = modify_landmark(modified, 'left_hip',
        y_delta=-round_amount * 0.3)
    modified = modify_landmark(modified, 'right_hip',
        y_delta=-round_amount * 0.3)
    return modified


def create_leaning_back(landmarks):
    """Torso leans backward — common with tight hamstrings."""
    modified = landmarks.copy()
    lean_amount = np.random.uniform(0.06, 0.14)
    back_shift  = np.random.uniform(0.02, 0.07)

    modified = modify_landmark(modified, 'nose',
        y_delta=-lean_amount * 0.8, x_delta=-back_shift)
    modified = modify_landmark(modified, 'left_shoulder',
        y_delta=-lean_amount, x_delta=-back_shift * 0.8)
    modified = modify_landmark(modified, 'right_shoulder',
        y_delta=-lean_amount, x_delta=-back_shift * 0.8)
    modified = modify_landmark(modified, 'left_elbow',
        y_delta=-lean_amount * 0.5, x_delta=-back_shift * 0.5)
    modified = modify_landmark(modified, 'right_elbow',
        y_delta=-lean_amount * 0.5, x_delta=-back_shift * 0.5)
    # Hips slide forward slightly
    modified = modify_landmark(modified, 'left_hip',
        x_delta=back_shift * 0.4)
    modified = modify_landmark(modified, 'right_hip',
        x_delta=back_shift * 0.4)
    return modified


def create_bent_knees(landmarks):
    """Knees not fully extended — knees rise up, ankles stay."""
    modified = landmarks.copy()
    bend_amount = np.random.uniform(0.06, 0.16)

    # Knees lift (y decreases = higher in frame)
    modified = modify_landmark(modified, 'left_knee',
        y_delta=-bend_amount)
    modified = modify_landmark(modified, 'right_knee',
        y_delta=-bend_amount)
    # Ankles shift slightly
    modified = modify_landmark(modified, 'left_ankle',
        y_delta=-bend_amount * 0.2)
    modified = modify_landmark(modified, 'right_ankle',
        y_delta=-bend_amount * 0.2)
    # Heels lift slightly
    modified = modify_landmark(modified, 'left_heel',
        y_delta=-bend_amount * 0.3)
    modified = modify_landmark(modified, 'right_heel',
        y_delta=-bend_amount * 0.3)
    return modified


def create_feet_not_flexed(landmarks):
    """Feet pointing down instead of flexed toward face."""
    modified = landmarks.copy()
    drop_amount = np.random.uniform(0.05, 0.12)

    # Foot index drops down (y increases), heel stays
    modified = modify_landmark(modified, 'left_foot_index',
        y_delta=drop_amount)
    modified = modify_landmark(modified, 'right_foot_index',
        y_delta=drop_amount)
    # Ankle shifts slightly
    modified = modify_landmark(modified, 'left_ankle',
        y_delta=drop_amount * 0.2)
    modified = modify_landmark(modified, 'right_ankle',
        y_delta=drop_amount * 0.2)
    return modified


def create_arms_not_pressing(landmarks):
    """Wrists float up or drift away from hips instead of pressing down."""
    modified = landmarks.copy()
    float_amount = np.random.uniform(0.05, 0.12)
    drift_amount = np.random.uniform(0.04, 0.10)

    if random.random() < 0.5:
        # Both arms float up
        modified = modify_landmark(modified, 'left_wrist',
            y_delta=-float_amount)
        modified = modify_landmark(modified, 'right_wrist',
            y_delta=-float_amount)
        modified = modify_landmark(modified, 'left_elbow',
            y_delta=-float_amount * 0.5)
        modified = modify_landmark(modified, 'right_elbow',
            y_delta=-float_amount * 0.5)
    else:
        # Arms drift sideways away from hips
        modified = modify_landmark(modified, 'left_wrist',
            x_delta=-drift_amount, y_delta=-float_amount * 0.3)
        modified = modify_landmark(modified, 'right_wrist',
            x_delta=drift_amount, y_delta=-float_amount * 0.3)
        modified = modify_landmark(modified, 'left_elbow',
            x_delta=-drift_amount * 0.5)
        modified = modify_landmark(modified, 'right_elbow',
            x_delta=drift_amount * 0.5)
    return modified



```

# Main function to create augumented dataset

```python

def augment_staff_dataset(input_csv, output_csv, samples_per_class=50):
    """
    Create augmented dataset with incorrect staff pose classes.
    """
    print(f"📊 Loading correct poses from {input_csv}...")
    df = pd.read_csv(input_csv)

    # Verify landmarks
    expected_cols = ['label']
    for lm in IMPORTANT_LANDMARKS:
        expected_cols.extend(get_landmark_columns(lm))

    missing_cols = [col for col in expected_cols if col not in df.columns]
    if missing_cols:
        print(f"⚠️  Warning: Missing columns: {missing_cols[:5]}...")

    # Filter correct poses
    df_correct = df[df['label'] == 'd'].copy()
    print(f"✅ Found {len(df_correct)} correct pose samples")

    augmentations = {
        'rounded_back':      create_rounded_back,
        'leaning_back':      create_leaning_back,
        'bent_knees':        create_bent_knees,
        'feet_not_flexed':   create_feet_not_flexed,
        'arms_not_pressing': create_arms_not_pressing,
    }

    all_samples = [df_correct]

    for label, aug_func in augmentations.items():
        print(f"🔄 Generating {samples_per_class} samples for class: {label}")
        augmented_samples = []

        for _ in range(samples_per_class):
            correct_sample = df_correct.sample(n=1).iloc[0].copy()
            landmark_cols  = [col for col in correct_sample.index if col != 'label']
            landmarks      = correct_sample[landmark_cols]

            modified_landmarks         = aug_func(landmarks)
            modified_landmarks['label'] = label
            augmented_samples.append(modified_landmarks)

        df_aug = pd.DataFrame(augmented_samples)
        all_samples.append(df_aug)
        print(f"   ✓ Created {len(df_aug)} samples")

    # Combine and shuffle
    df_final = pd.concat(all_samples, ignore_index=True)
    df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)

    df_final.to_csv(output_csv, index=False)

    print(f"\n✅ Augmented dataset saved to {output_csv}")
    print(f"\n📈 Class Distribution:")
    class_counts = df_final['label'].value_counts().sort_index()
    for label, count in class_counts.items():
        print(f"   {label:25s}: {count:4d} samples")
    print(f"\n{'='*50}")
    print(f"Total samples:          {len(df_final)}")
    print(f"Total classes:          {df_final['label'].nunique()}")
    print(f"Features per sample:    {len([col for col in df_final.columns if col != 'label'])}")

    return df_final
```
